```python
duplicate_rows = data.duplicated().sum()

print("Duplicate rows:", duplicate_rows)
```

```python
invalid_flags = data[~data["FLAG"].isin([0, 1])]

print("Invalid FLAG values:", len(invalid_flags))
```

```python
missing_total = data.isna().sum().sum()

print("Total missing values:", missing_total)
```

```python
consumption_cols = data.columns[1:-1]

data["missing_ratio"] = data[consumption_cols].isna().mean(axis=1)

data["missing_ratio"].describe()
```

```python
negative_count = (data[consumption_cols] < 0).sum().sum()

print("Negative consumption values:", negative_count)
```

```python
data.drop(
    columns=["missing_ratio"],
    inplace=True
)
```

```python
print("Shape:", data.shape)

print("\nDuplicate rows:", data.duplicated().sum())

print("\nDuplicate customers:", data["CONS_NO"].duplicated().sum())

print("\nFLAG:")
print(data["FLAG"].value_counts(dropna=False))

print("\nMissing values:", data.isna().sum().sum())
```

```python
data.to_csv(
    "../data/processed/cleaned_electricity_data.csv",
    index=False
)
```
